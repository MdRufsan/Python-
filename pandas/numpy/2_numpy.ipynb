**Indexing and sliceing**

```python
# 1D array indexing

import numpy as np
a=np.array([1,2,3,4,5])
print(a[0])
print(a[-1])
```

```python
# 1D array slicing

import numpy as np
a=np.array([1,2,3,4,5])
print(a[0:3])
print(a[2:])
```

```python
# 2D array indexing
a=np.array([[1,2,3],[4,5,6]])
print(a[0,1]) # 1st row 2nd col
print(a[1,2]) # 2nd row 3rd col
```

```python
# 2D array slicing

a=np.array([[1,2,3],[4,5,6]])
print(a[:,1]) # all rows 2nd col
print(a[0,:]) # 1st row's all cols
```

```python
# boolean indexing (important for filtering)

import numpy as np
a=np.array([1,2,3,4,5,6])
print(a[a>2])

print(a[a%2==0])
print(a[a%2==1])
```

**Array operations**

```python
# arithmetic operations

import numpy as np
a=np.array([1,2,3])
b=np.array([4,5,6])

print(a+b)
print(a-b)
print(a*b)
print(a/b)
```

```python
# scalar operations

import numpy as np
a=np.array([1,2,3])
print(a+2)
print(a-2)
print(a*2)
print(a/2)
```

```python
#matrix multiplication (dot product)

import numpy as np
a=np.array([[1,2],[3,4]])
b=np.array([[5,6],[7,8]])
print(a.dot(b))
print(np.dot(a,b))
```

```python
# eliment wise square,power,root

import numpy as np
a=np.array([1,2,3])
print(np.square(a))
print(np.power(a,2))
print(np.sqrt(a))
```

```python
# aggregate functions

import numpy as np
a=np.array([1,2,3,4,5])
print(np.sum(a))
print(np.mean(a))
print(np.min(a))
print(np.max(a))
print(np.std(a))
print(np.var(a))
print(np.median(a))
print(np.percentile(a,25))
print(np.percentile(a,50))
print(np.percentile(a,75))
print(np.percentile(a,100))
print(np.percentile(a,0))
print(np.percentile(a,5))
```

Missing values ,boolean masking & filtering

```python
# MSSING values(np.nan --> not a number)

import numpy as np

arr=np.array([1,2,3,np.nan,5])
print(arr)
print(np.isnan(arr)) # NaN check --> isnan()
```

```python
# clean NaN

import numpy as np

arr=np.array([1,2,3,np.nan,5])
print(arr)

cleaned = arr[~np.isnan(arr)] # without NaN

print(cleaned)
```

```python
# find mean if remains NaN

import numpy as np

arr=np.array([1,2,3,np.nan,5])
print(arr)

mean=np.nanmean(arr) # mean without NaN values Mean=(1+2+3+5)/4=2.75
print(mean)
```

```python
# boolean masking & filtering

arr=np.array([10,20,30,5,50])
mask= arr >= 20
print(mask)

filtered_arr=arr[mask]
print(filtered_arr)
```

```python
arr=np.array([10,20,30,5,50])
print('Between 20 & 40 : ', arr[(arr > 20) & (arr < 40)])
print('Not between 20 & 40 : ', arr[~(arr > 20) & (arr < 40)])
```

```python
# 2d arrary filtering

a=np.array([[1,2,3],[4,5,6]])
print(a)
mask2d= a > 2
print(mask2d)
filtered_arr=a[mask2d]
print(filtered_arr)
```

**Stacking**(eksathe jora lagano)

```python
# vstack --> vertically(up-down)
import numpy as np

a=np.array([1,2,3])
b=np.array([4,5,6])

vstacked=np.vstack((a,b))
print(vstacked)
```

```python
# hstack --> horizontally
import numpy as np

a=np.array([1,2,3])
b=np.array([4,5,6])

hstacked=np.hstack((a,b))
print(hstacked)
```

```python
#stack --> will make 3D (by custimming axis)

a=np.array([1,2,3])
b=np.array([4,5,6])

stacked=np.stack((a,b),axis=0)
print(stacked)

print('...........................................')
stacked2=np.stack((a,b),axis=1)
print(stacked2)
```

```python
# splitting(vag kora)
a=np.array([1,2,3,4,5,6])
print(np.split(a,3))
```

```python
# 2D array split

a=np.array([[1,2,3,4],[5,6,7,8]])
print("Original array:")
print(a)
print('.............')
print("Horizontal split into 2:")
print(np.hsplit(a,2)) # Split into 2 columns
print('.............')
print("Vertical split into 2:")
print(np.vsplit(a,2)) # Split into 2 rows
```
