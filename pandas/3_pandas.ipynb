**Data Cleaning with pandas**

ডেটাতে যেসব ভুল বা সমস্যা থাকে সেগুলো ঠিক করা — যেমন missing value, duplicate row, column type ইত্যাদি।

```python
import pandas as pd

# Raw data creation (missing value ও duplicate সহ)
data = {
    'Name': ['Shanto', 'Rafi', 'Nabila', 'Rafi'],
    'Age': [25, None, 23, None],
    'Score': [85, 90, None, 90]
}
df = pd.DataFrame(data)

display(df)
```

```python
#  Missing value check

print(df.isnull())
print(df.isnull().sum())  ## কোন column-এ কয়টা missing value
```

```python
#Missing পূরণ করো (fillna)

df['Age'] = df['Age'].fillna(df['Age'].mean())

df['Score']= df['Score'].fillna(0)

df
```

```python
# find duplicate

print(df.duplicated())               # কোন row টা duplicate

df = df.drop_duplicates()           # duplicate row বাদ
df
```

```python
#  ডেটা টাইপ ঠিক করো:


df['Age'] = df['Age'].astype(int)
print(df.dtypes)
```

```python
 # clean dataframe
print(df)
```

**Data Manipulation in pandas**
 কিভাবে ডেটাকে সাজানো, গ্রুপ করা ও বিশ্লেষণ করা যায়।

```python
import pandas as pd

data = {
    'Name': ['Shanto', 'Rafi', 'Nabila', 'Asif', 'Tania'],
    'Dept': ['SWE', 'SWE', 'BBA', 'BBA', 'SWE'],
    'Score': [85, 90, 88, 82, 92]
}

df = pd.DataFrame(data)
df
```

```python
# Data Sorting:

# Score অনুসারে বড় থেকে ছোট (descending):

sorted_df = df.sort_values(by='Score', ascending=False)
print(sorted_df)
print('................................')
# Score অনুসারে ছোট থেকে বড় (ascending):
sorted_df = df.sort_values(by='Score', ascending=True)
print(sorted_df)
```

```python
#নাম অনুযায়ী (A-Z):


print(df.sort_values(by='Name'))
```

Grouping & Aggregation:

```python
# প্রতিটি Dept-এর মোট Score:


print(df.groupby('Dept')['Score'].sum())

 #প্রতিটি Dept-এর গড় স্কোর:
print(df.groupby('Dept')['Score'].mean())
```

```python
#প্রতিটি Dept-এ কয়জন ছাত্র আছে:


print(df['Dept'].value_counts())
# অথবা
print(df.groupby('Dept')['Name'].count())
```

```python
#কোন Dept-এ সবচেয়ে বেশি ছাত্র আছে?


dept_count = df['Dept'].value_counts()
print(" Number of Students per Dept:\n", dept_count)

# সবচেয়ে বেশি ছাত্র কোন Dept-এ
most_students_dept = dept_count.idxmax()
print('সবচেয়ে বেশি ছাত্র ',most_students_dept, ' Dept-এ থাকে')
```

```python
import pandas as pd

data = {
    'Name': ['Shanto', 'Rafi', 'Nabila', 'Asif', 'Tania'],
    'Dept': ['SWE', 'SWE', 'BBA', 'BBA', 'SWE'],
    'Score1': [85, 90, 88, 82, 92],
    'Score2': [80, 91, 85, 79, 95]
}

df = pd.DataFrame(data)
df
```

```python
df_sorted = df.sort_values(by='Score1', ascending=False)
print(df_sorted)
```

```python
df_sorted = df.sort_values(by='Score2', ascending=False)
print(df_sorted)
```

```python
#Dept অনুযায়ী গ্রুপ করে প্রতিটি বিভাগের সব স্কোর দেখাও:

grouped = df.groupby('Dept')
print(grouped[['Score1', 'Score2']].mean())
```

```python
 #Aggregation

 # প্রতিটি বিভাগের মধ্যে:গড় স্কোর,সর্বোচ্চ স্কোর,মোট ছাত্র সংখ্যা


agg_result = df.groupby('Dept').agg({
    'Score1': ['mean', 'max'],
    'Name': 'count'
})

print(agg_result)
```

```python
#নতুন Column তৈরি

#Score1 ও Score2 এর গড় নিয়ে Average column তৈরি করো:

df['Average'] = (df['Score1'] + df['Score2']) / 2
print(df)
```

```python
#Grade Column তৈরি করো:

def assign_grade(avg):
    if avg >= 90:
        return 'A+'
    elif avg >= 80:
        return 'A'
    else:
        return 'B'

df['Grade'] = df['Average'].apply(assign_grade)
print(df)
```

**Merging & Joining in pandas**

pandas-এ তিনটি প্রধান উপায় আছে DataFrame একত্র করার:

Concatenate	pd.concat()	একের পর এক যোগ করা (row/column-wise)
Merge	pd.merge()	SQL-style অনুযায়ী একাধিক DataFrame যুক্ত করা
Join	df.join()	index বা কলাম অনুযায়ী যুক্ত করা

```python
#. pd.concat() → একসাথে যোগ করা

# Example: Row-wise concatenate

import pandas as pd

df1 = pd.DataFrame({
    'Name': ['Shanto', 'Rafi'],
    'Dept': ['SWE', 'SWE']
})

df2= pd.DataFrame({
    'Name' : ['Shawon','Hasib'],
    'Dept' : ['BBA','BBA']
})


result = pd.concat([df1, df2])
print(result)
```

```python
# Column-wise concat:

df3 = pd.DataFrame({
    'Name': ['Shanto', 'Rafi'],
    'Score': [85, 90]
})

df4 = pd.DataFrame({
    'Grade': ['A', 'A+']
})

combined = pd.concat([df3, df4],axis=1)
print(combined)
```

```python
 #pd.merge() → SQL-style Merge

#Example: Merge on common column (like foreign key)


students = pd.DataFrame({
    'ID': [1, 2, 3],
    'Name': ['Shanto', 'Rafi', 'Tania']
})

scores = pd.DataFrame({
    'ID': [1, 2, 3],
    'Score': [85, 90, 92]
})

grade=pd.DataFrame({
    'ID':[1, 2, 3],
    'Grade':['A','A+','A+']
})



merged = pd.merge(students, scores, on='ID')
print(merged)
```

```python
#Merge with different types (left, right, outer, inner):


# inner merge (default): শুধু common ID
pd.merge(students, scores, on='ID', how='inner')
```

```python
# left join: students-এর সব data রাখবে
pd.merge(students, scores, on='ID', how='left')
```

```python
# df.join() → Index অনুযায়ী একত্র


df_a = pd.DataFrame({
    'Name': ['Santo', 'Rafi'],
    'Score': [85, 90]
}, index=['a', 'b'])

df_b = pd.DataFrame({
    'Grade': ['A', 'A+']
}, index=['a', 'b'])

joined = df_a.join(df_b)
print(joined)
```
