```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# "tips" ডেটাসেট লোড করা
tips = sns.load_dataset("tips")

# প্রথম কিছু সারি দেখা
print(tips.head(10))
```

```python
#Scatter Plot
sns.scatterplot(x="total_bill", y="tip", data=tips,color='red')
plt.title("Total Bill vs Tip")
plt.show()
```

```python
#Line Plot

sns.lineplot(x="size", y="total_bill", data=tips,color='green')
plt.title("Size vs Total Bill")
plt.show()
```

```python
#Bar Plot

sns.barplot(x="day", y="total_bill", data=tips,color='orange')
plt.title("Average Total Bill by Day")
plt.show()
```

```python
#Count Plot (একটি ক্যাটেগরির সংখ্যা)
sns.countplot(x="day", data=tips,color='blue')
plt.title("Count of Customers by Day")
plt.show()
```

```python
#Histogram

sns.histplot(data=tips, x="total_bill", bins=20, kde=True,color='green')
plt.title("Distribution of Total Bill")
plt.show()
```

```python
#Box Plot

sns.boxplot(x="day", y="total_bill", data=tips,color='skyblue')
plt.title("Boxplot of Total Bill by Day")
plt.show()
```

sns.heatmap(corr, annot=True, cmap="coolwarm")
👉 এই লাইনে Heatmap আঁকা হচ্ছে।

ব্যাখ্যা:

corr → আমরা যেই Correlation Matrix বানিয়েছি সেটা প্লট করা হচ্ছে।

annot=True → প্রতিটি cell-এর ভিতরে সম্পর্কের মান (correlation value) লিখে দেখাবে।

cmap="coolwarm" → রঙের প্যালেট (নীল থেকে লাল) ব্যবহার করা হচ্ছে।

🎯 অর্থাৎ, বড় পজিটিভ সম্পর্ক লাল, বড় নেগেটিভ সম্পর্ক নীল, মাঝামাঝি মান সাদা বা হালকা রঙে।

```python
#Correlation Matrix + Heatmap

# Iris ডেটাসেট লোড
iris = sns.load_dataset("iris")
corr = iris.corr(numeric_only=True)

# হিটম্যাপ আঁকা
sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap of Iris Dataset")
plt.show()
```

```python
# গ্রাফ কাস্টমাইজেশন (রং, স্টাইল, মার্কার)

sns.set_style("whitegrid")

sns.scatterplot(x="total_bill", y="tip", hue="sex", style="smoker", data=tips)
plt.title("Tips by Gender and Smoking Status")
plt.show()
```

```python
#সাবপ্লট (একসাথে একাধিক গ্রাফ)

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

sns.boxplot(x="day", y="total_bill", data=tips, ax=axes[0],color='green')
axes[0].set_title("Boxplot")

sns.barplot(x="day", y="tip", data=tips, ax=axes[1],color='hotpink')
axes[1].set_title("Barplot")

plt.tight_layout()
plt.show()
```

Common seaborn plots:

sns.histplot() → Histogram

sns.boxplot() → Box & outlier

sns.countplot() → category-wise count

sns.barplot() → Mean with error bar

sns.heatmap() → Matrix plot (correlation, etc.)

sns.pairplot() → Multi-variable relationships
