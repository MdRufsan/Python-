**Matplotlib দিয়ে Basic Line Plot**

💡 কি হলো?

plt.plot(x, y) → x ও y এর মান দিয়ে রেখা আঁকা হলো।

plt.title() → গ্রাফের নাম।

plt.xlabel(), plt.ylabel() → অক্ষের লেবেল।

plt.grid(True) → গ্রিড দেখায়, যা পড়তে সাহায্য করে।

plt.show() → প্লট দেখানোর কমান্ড।

```python
import matplotlib.pyplot as plt

# Data
x = [1, 2, 3, 4, 5]
y = [10, 15, 13, 18, 16]

# Line Plot
plt.plot(x, y,color='red')

# Title and labels
plt.title('Basic Line Plot', color='red')
plt.xlabel('<-- X axis -->', color='blue')
plt.ylabel('<-- Y axis -->', color='blue')

# Show grid
plt.grid(True, color='green')

# Show plot
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

xpoints = np.array([1, 2, 6, 8])
ypoints = np.array([3, 8, 1, 10])

plt.plot(xpoints, ypoints,color='hotpink',linewidth = '20.5')

plt.grid(True)
plt.show()
```

**Display Multiple Plots**
With the subplot() function you can draw multiple plots in one figure:

```python
import matplotlib.pyplot as plt
import numpy as np

# Plot 1:
x1 = np.array([0, 1, 2, 3])
y1 = np.array([3, 8, 1, 10])

plt.subplot(1, 3, 1) # 1 row, 3 columns, position 1
plt.plot(x1, y1, color='red')
plt.title('Plot 1')

# Plot 2:
x2 = np.array([0, 1, 2, 3])
y2 = np.array([10, 20, 30, 40])

plt.subplot(1, 3, 2) # 1 row, 3 columns, position 2
plt.plot(x2, y2, color='blue')
plt.title('Plot 2')

# Plot 3:
x3 = np.array([0, 1, 2, 3])
y3 = np.array([5, 12, 8, 15])

plt.subplot(1, 3, 3) # 1 row, 3 columns, position 3
plt.plot(x3, y3, color='green')
plt.title('Plot 3')

plt.tight_layout() # Adjust layout to prevent overlapping titles/labels
plt.show()
```

**Scatter Plot (বিন্দু দ্বারা সম্পর্ক বোঝানো)**

📌 ব্যাখ্যা:

plt.scatter() → scatter plot তৈরি করে

marker='o' → বৃত্ত চিহ্ন ব্যবহার

এর মাধ্যমে দুইটি ভ্যারিয়েবলের সম্পর্ক দেখা যায় (correlation বোঝা যায়)

```python
import matplotlib.pyplot as plt

# Data
x = [1, 2, 3, 4, 5]
y = [5, 7, 4, 9, 6]

# Scatter plot
plt.scatter(x, y, color='green', marker='o')

plt.title('Scatter Plot Example')
plt.xlabel('X values')
plt.ylabel('Y values')
plt.grid(True)
plt.show()
```

**Create Histogram**
In Matplotlib, we use the hist() function to create histograms.

The hist() function will use an array of numbers to create a histogram, the array is sent into the function as an argument.

For simplicity we use NumPy to randomly generate an array with 250 values, where the values will concentrate around 170, and the standard deviation is 10. Learn more about Normal Data Distribution in our Machine Learning Tutorial.

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.random.normal(170, 10, 250)

plt.hist(x,color='pink')
plt.show()
```

```python
import matplotlib.pyplot as plt

# Data
ages = [22, 25, 28, 23, 22, 27, 30, 35, 40, 42, 41, 39, 38, 45, 50]

# Histogram plot
plt.hist(ages, bins=5, color='purple', edgecolor='black')

plt.title('Age Distribution')
plt.xlabel('Age Range')
plt.ylabel('Count')
plt.show()
```

**Pie Chart (বিভাগভিত্তিক ভাগ দেখানো)**
📌 ব্যাখ্যা:

autopct='%1.1f%%' → শতকরা হারে মান দেখায়

startangle=140 → শুরু ঘুরানোর অ্যাঙ্গেল

plt.axis('equal') → সার্কেল রাখতে সাহায্য করে

```python
# Data
labels = ['SWE', 'BBA', 'ENG', 'LAW']
sizes = [40, 25, 20, 15]

# Pie chart
plt.pie(sizes, labels=labels, autopct='%1.1f%%', startangle=140)

plt.title('Department-wise Student Distribution')
plt.axis('equal')  # Circle shape
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

y = np.array([35, 25, 25, 15])
mylabels = ["Apples", "Bananas", "Cherries", "Dates"]
mycolors = ["black", "hotpink", "b", "#4CAF50"]

plt.pie(y, labels = mylabels, colors = mycolors,autopct='%1.1f%%', startangle=140)
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

y = np.array([35, 25, 25, 15])
mylabels = ["Apples", "Bananas", "Cherries", "Dates"]

plt.pie(y, labels = mylabels)
plt.legend(title = "Four Fruits:")
plt.show()
```

** Box Plot (Outlier ও Median)**
🔹 মিডিয়ান, কিউটাইল, outlier – সব দেখা যায়
🔹 Box plot স্ট্যাটিস্টিক্যাল visualization-এর জন্য খুব গুরুত্বপূর্ণ

```python
data = [12, 15, 13, 17, 19, 14, 12, 15, 16, 200]  # 200 is an outlier

plt.boxplot(data)
plt.title('Box Plot Example')
plt.ylabel('Values')
plt.show()
```

** Stack Plot (Layered Trend)**

```python
# Data
days = [1, 2, 3, 4, 5]
eating = [2, 3, 4, 3, 2]
sleeping = [7, 8, 6, 7, 8]
working = [8, 7, 8, 9, 8]

# Stack Plot
plt.stackplot(days, eating, sleeping, working, labels=['Eat', 'Sleep', 'Work'], colors=['#ff9999','#66b3ff','#99ff99'])
plt.legend()
plt.title('Daily Activities')
plt.xlabel('Day')
plt.ylabel('Hours')

plt.show()
```

```python
#Fill Between (Highlight range)

x = [1, 2, 3, 4, 5]
y = [2, 3, 5, 4, 6]

plt.plot(x, y, color='green')
plt.fill_between(x, y, color='lightgreen', alpha=0.5)

plt.title('Fill Between Example')
plt.show()
```
